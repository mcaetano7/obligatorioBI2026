```python
from pathlib import Path
import pandas as pd

DATOS = Path("../datos")

# Ver qué CSV hay
for f in sorted(DATOS.rglob("*.csv")):
    print(f.name)
```

```python
for f in sorted(DATOS.rglob("*.csv")):
    df = pd.read_csv(f)
    print(f"\n=== {f.name} ===")
    print(df.shape)
    df.info()
    print("Nulos:", df.isna().sum().sum(), "| Duplicados:", df.duplicated().sum())
```
